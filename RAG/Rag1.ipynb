## Chunking strategies IMPORTANT

#### Size Based -> Divisao do texto em tamanhos iguais - Palavras cortadas e contexto fraco

- Overlap de chunks seguintes para não cortar, (vai ter algum texto duplicado)

#### Structure Based -> Dividir o texto com base nos paragrafos/secções
 
#### Semantic Based -> Divide o texto em grupos com frases / secções relacionadas
